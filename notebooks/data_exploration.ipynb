## 1. Import libraries

```python
import pandas as pd
import requests
import time

print("Libraries imported successfully!")
```

## 2. Configure the OWID API URL

```python
url = "https://ourworldindata.org/grapher/international-tourist-trips.csv?v=1&csvType=full&useColumnShortNames=false"

print("Tourism dataset URL configured successfully!")
print(url)
```

## 3. Extract tourism data

```python
response = requests.get(url, timeout=120)

response.raise_for_status()

from io import StringIO

tourism_df = pd.read_csv(StringIO(response.text))

print("Tourism data extracted successfully!")
print("Total records:", len(tourism_df))
```

## 4. Display dataset overview

```python
print("Dataset Shape:", tourism_df.shape)

print("\nColumn Names:")
print(tourism_df.columns.tolist())

print("\nFirst 5 Records:")
display(tourism_df.head())
```

## 5. Check missing values

```python
print("Missing Values in Each Column:")

missing_values = tourism_df.isnull().sum()

display(
    missing_values.reset_index(
        name="missing_count"
    ).rename(columns={"index": "column_name"})
)
```

## 6. Check duplicate records

```python
duplicate_count = tourism_df.duplicated().sum()

print("Total Duplicate Records:", duplicate_count)
```

## 7. Generate statistical summary

```python
print("Statistical Summary of Tourism Dataset:")

display(tourism_df.describe())
```

## 8. Analyze tourism records by country

```python
country_column = "Entity"

country_analysis = (
    tourism_df[country_column]
    .value_counts()
    .reset_index()
)

country_analysis.columns = ["country", "record_count"]

print("Top 10 Countries/Entities by Record Count:")

display(country_analysis.head(10))
```

## 9. Analyze tourism records by year

```python
year_analysis = (
    tourism_df["Year"]
    .value_counts()
    .sort_index()
    .reset_index()
)

year_analysis.columns = ["year", "record_count"]

print("Tourism Records by Year:")

display(year_analysis)
```

## 10. Find the latest year with available tourism data

```python
latest_year = tourism_df["Year"].max()

print("Latest Year Available:", latest_year)
```

## 11. Identify the top 10 countries/entities by tourism records

```python
top_10_countries = (
    tourism_df["Entity"]
    .value_counts()
    .head(10)
    .reset_index()
)

top_10_countries.columns = ["country", "record_count"]

print("Top 10 Countries/Entities:")

display(top_10_countries)
```

## 12. Analyze tourism trends over time

```python
yearly_trends = (
    tourism_df.groupby("Year")["Arrivals of tourists from abroad"]
    .sum()
    .reset_index()
)

yearly_trends = yearly_trends.sort_values("Year")

print("Year-wise Tourism Trends:")

display(yearly_trends)
```

## 13. Analyze tourism arrivals by world region

```python
region_analysis = (
    tourism_df.groupby("World region according to OWID")[
        "Arrivals of tourists from abroad"
    ]
    .sum()
    .reset_index()
)

region_analysis.columns = ["region", "total_arrivals"]

region_analysis = region_analysis.sort_values(
    "total_arrivals", ascending=False
)

print("Tourism Arrivals by World Region:")

display(region_analysis)
```

## 14. Analyze tourism arrivals by country

```python
country_arrivals = (
    tourism_df.groupby("Entity")[
        "Arrivals of tourists from abroad"
    ]
    .sum()
    .reset_index()
)

country_arrivals.columns = ["country", "total_arrivals"]

country_arrivals = country_arrivals.sort_values(
    "total_arrivals", ascending=False
)

print("Top 10 Countries/Entities by Total Tourist Arrivals:")

display(country_arrivals.head(10))
```

## 15. Visualize tourism arrivals over time

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))

plt.plot(
    yearly_trends["Year"],
    yearly_trends["Arrivals of tourists from abroad"],
    marker="o"
)

plt.title("International Tourist Arrivals Over Time")
plt.xlabel("Year")
plt.ylabel("Tourist Arrivals")
plt.grid(True)

plt.show()
```

## 16. Final data exploration summary

```python
print("GLOBAL TOURISM INTELLIGENCE")
print("=" * 45)

print("Total Records:", len(tourism_df))
print("Total Columns:", len(tourism_df.columns))

print("Unique Countries/Entities:", tourism_df["Entity"].nunique())

print("Year Range:",
      tourism_df["Year"].min(),
      "to",
      tourism_df["Year"].max())

print("Missing Values:", tourism_df.isnull().sum().sum())

print("Duplicate Records:", tourism_df.duplicated().sum())

print("Latest Year Available:", tourism_df["Year"].max())

print("=" * 45)
print("DATA EXPLORATION COMPLETED SUCCESSFULLY!")
```
